```python
import json
import numpy as np
```

```python
bible = json.load(open('data/segond_1910.json', 'r'))
```

```python
gospels = {"Matthieu", "Marc", "Luc", "Jean"}
versets = np.array([v for v in bible["verses"] if v["book_name"] in gospels])
```

```python
for verset in versets[:5]:
    print(verset["book_name"], verset["text"])
```
